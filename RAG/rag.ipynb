```python
## Data Ingestion [data -> chunks -> embeds]
from langchain_community.document_loaders import TextLoader
loader = TextLoader("speech.txt")
txt_doc = loader.load()
txt_doc
```

```python
## document to documents
import os
from langchain_text_splitters import RecursiveCharacterTextSplitter
txt_split = RecursiveCharacterTextSplitter(chunk_size= 500, chunk_overlap= 30)
doc=txt_split.split_documents(txt_doc)
doc[:2]
```

```python
## Vector embedding and Vector store
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import Chroma
db = Chroma.from_documents(doc, OllamaEmbeddings(model="llama3:latest"))
```

```python
## Vector Database
query = "When was the speech delivered"
result = db.similarity_search(query)
result[0].page_content
```

```python
## FAISS Database
from langchain_community.vectorstores import FAISS
db1 = FAISS.from_documents(doc,OllamaEmbeddings(model="llama3:latest"))
```

```python
query = "What is the main idea of the speech?"
result = db1.similarity_search(query)
result
```
